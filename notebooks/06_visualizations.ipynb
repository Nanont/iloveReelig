```python
import pandas as pd
import joblib
import os

# 1) Load the same df_ml_base you used for training
df_base = pd.read_csv("../data_processed/df_ml_base.csv")
print("df_base shape:", df_base.shape)
print("df_base columns:", df_base.columns)

# 2) Recreate time features from timestamp (same as in 05_ml_model)
df_base["timestamp"] = pd.to_datetime(df_base["timestamp"], errors="coerce")
df_base["hour"] = df_base["timestamp"].dt.hour
df_base["weekday"] = df_base["timestamp"].dt.weekday
df_base["month"] = df_base["timestamp"].dt.month

# 3) Now build X_all with the same feature columns used in training
feature_cols = [
    "type",
    "district",
    "province",
    "hour",
    "weekday",
    "month",
    "lat",
    "lon",
]

# sanity check: make sure all exist
for c in feature_cols:
    assert c in df_base.columns, f"Missing column in df_base: {c}"

X_all = df_base[feature_cols].copy()
print("X_all shape:", X_all.shape)

# 4) Load best model
best_model = joblib.load("../ML/models/best_efficiency_model_knn.pkl")  # make sure model_path is set correctly

# 5) Predict efficiency_score
y_pred_score = best_model.predict(X_all)

# 6) Attach predictions to df_base
df_base["pred_efficiency_score"] = y_pred_score

# 7) Save for Power BI
os.makedirs("../data_processed", exist_ok=True)
out_path = "../data_processed/for_powerbi_with_efficiency_score.csv"
df_base.to_csv(out_path, index=False)

print("Saved:", out_path, "with shape:", df_base.shape)
```
