# 02_CLEAN — Traffy Fondue Data Preperation

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load data
df = pd.read_csv('../data_raw/bangkok_traffy_raw.csv', low_memory=False)
df_ex1 = pd.read_csv('data_raw/external_raw/flood_30days_gistda_formatted.csv', low_memory=False)
df_ex2 = pd.read_csv('data_raw/external_raw/flood_risk!!!.csv', low_memory=False)
df_ex3 = pd.read_csv('data_raw/external_raw/pm25_api_full.csv', low_memory=False)
df_ex4 = pd.read_csv('data_raw/external_raw/risk_path.csv', low_memory=False)

df.head()
```

# Standardize Missing Values

```python
missing_tokens = ["", " ", "N/A", "n/a", "na", "null", "None", "-", "ไม่ระบุ", "{}"]
df = df.replace(missing_tokens, np.nan)
df.head(10)
```

# coordinate split
จาก dataset คาดว่า latitude longitude สลับกัน

```python
def split_coords(coords):
    if pd.isna(coords) or not isinstance(coords, str) or ',' not in coords:
        return pd.Series([np.nan, np.nan])
    lat, lon = coords.split(',', 1)
    try:
        return pd.Series([float(lon), float(lat)])
    except:
        return pd.Series([np.nan, np.nan])

df[['lat', 'lon']] = df['coords'].apply(split_coords)
 
# Remove invalid Thailand coordinates
df.loc[(df.lat < 5) | (df.lat > 21), ['lat']] = np.nan
df.loc[(df.lon < 97) | (df.lon > 106), ['lon']] = np.nan

df[['lat', 'lon']].head(20)
```

## Fill count_reopen Missing

```python
df['count_reopen'] = df['count_reopen'].fillna(0).astype(int)
df['count_reopen'].value_counts().head()
```

## Timestamp Cleaning

```python
df['timestamp'] = pd.to_datetime(df['timestamp'], errors='coerce')
df['last_activity'] = pd.to_datetime(df['last_activity'], errors='coerce')

# For missing last_activity, use timestamp
df['last_activity'] = df['last_activity'].fillna(df['timestamp'])

df[['timestamp','last_activity']].head()
```

# column-by-column missing value

```python
df['type'] = df['type'].fillna('Unknown')
# Fill using mode per type
df['type'] = df['type'].fillna('{Unassigned}')
df['comment'] = df['comment'].fillna('')
df['has_photo'] = df['photo'].notna().astype(int)
df['has_photo_after'] = df['photo_after'].notna().astype(int)
df['star'] = df['star'].fillna(-1).astype(int)
df['last_activity'] = df['last_activity'].fillna(df['timestamp'])
df['state'] = df['state'].fillna('unknown')
df['count_reopen'] = df['count_reopen'].fillna(0).astype(int)
```

# Handle Missing Value of district and subdistrict

```python
# ---------------------------------------------------------
# Extract district (เขต…) and subdistrict (แขวง…) from Thai address
# Validate extracted values against original
# Drop rows where they conflict
# Fill missing afterward
# ---------------------------------------------------------

import re
import numpy as np
import pandas as pd

# ---- Helper extractors -----------------------------------------------------

def extract_district_th(addr):
    """Extract district name after 'เขต' from Thai address."""
    if pd.isna(addr):
        return np.nan
    match = re.search(r"เขต\s*([ก-๙]+)", str(addr))
    return match.group(1).strip() if match else np.nan


def extract_subdistrict_th(addr):
    """Extract subdistrict name after 'แขวง' from Thai address."""
    if pd.isna(addr):
        return np.nan
    match = re.search(r"แขวง\s*([ก-๙]+)", str(addr))
    return match.group(1).strip() if match else np.nan


# ---- Extract from address (temporary columns) ------------------------------

df["district_from_addr"] = df["address"].apply(extract_district_th)
df["subdistrict_from_addr"] = df["address"].apply(extract_subdistrict_th)


# ---- Normalize spacing for comparison --------------------------------------

def clean_text(x):
    if pd.isna(x):
        return x
    return str(x).replace(" ", "").strip()

df["district"] = df["district"].apply(clean_text)
df["subdistrict"] = df["subdistrict"].apply(clean_text)
df["district_from_addr"] = df["district_from_addr"].apply(clean_text)
df["subdistrict_from_addr"] = df["subdistrict_from_addr"].apply(clean_text)


# ---- Drop rows where extracted values conflict with existing ones ----------

conflict_district = (
    df["district"].notna() &
    df["district_from_addr"].notna() &
    (df["district"] != df["district_from_addr"])
)

conflict_subdistrict = (
    df["subdistrict"].notna() &
    df["subdistrict_from_addr"].notna() &
    (df["subdistrict"] != df["subdistrict_from_addr"])
)

conflict_rows = conflict_district | conflict_subdistrict

print("Dropping", conflict_rows.sum(), "rows due to conflicting district/subdistrict.")

df = df[~conflict_rows].copy()


# ---- Fill missing district/subdistrict using extracted values --------------

df.loc[df["district"].isna() | (df["district"] == ""), "district"] = df["district_from_addr"]
df.loc[df["subdistrict"].isna() | (df["subdistrict"] == ""), "subdistrict"] = df["subdistrict_from_addr"]


# ---- Remove temporary columns ----------------------------------------------

df.drop(columns=["district_from_addr", "subdistrict_from_addr"], inplace=True)
```

## Feature Engineering

# fix or remove impossible broken rows

```python
print("Dataset shape before cleaning:", df.shape)
df = df.dropna(subset=['ticket_id', 'timestamp', 'lat', 'lon', 'organization', 'address'])
df = df[df['province'].str.lower() == 'กรุงเทพมหานคร'].copy()
df = df[df['count_reopen'].fillna(0) >= 0]
df = df.dropna(how='all')
df = df.reset_index(drop=True)
print("Dataset shape after cleaning:", df.shape)
print("Sample rows:")
df.head(10)
```
