# 02_CLEAN — Traffy Fondue Data Preperation

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load data
df = pd.read_csv('../data_raw/bangkok_traffy_raw.csv', low_memory=False)

df.head()
```

# Standardize Missing Values

```python
missing_tokens = ["", " ", "N/A", "n/a", "na", "null", "None", "-", "ไม่ระบุ", "{}"]
df = df.replace(missing_tokens, np.nan)
df.head(10)
```

# coordinate split
จาก dataset คาดว่า latitude longitude สลับกัน

```python
def split_coords(coords):
    if pd.isna(coords) or not isinstance(coords, str) or ',' not in coords:
        return pd.Series([np.nan, np.nan])
    lat, lon = coords.split(',', 1)
    try:
        return pd.Series([float(lon), float(lat)])
    except:
        return pd.Series([np.nan, np.nan])

df[['lat', 'lon']] = df['coords'].apply(split_coords)
 
# Remove invalid Thailand coordinates
df.loc[(df.lat < 5) | (df.lat > 21), ['lat']] = np.nan
df.loc[(df.lon < 97) | (df.lon > 106), ['lon']] = np.nan

df[['lat', 'lon']].head(20)
```

## Fill count_reopen Missing

```python
df['count_reopen'] = df['count_reopen'].fillna(0).astype(int)
df['count_reopen'].value_counts().head()
```

## Timestamp Cleaning

```python
df['timestamp'] = pd.to_datetime(df['timestamp'], errors='coerce')
df['last_activity'] = pd.to_datetime(df['last_activity'], errors='coerce')

# For missing last_activity, use timestamp
df['last_activity'] = df['last_activity'].fillna(df['timestamp'])

df[['timestamp','last_activity']].head()
```

# column-by-column missing value

```python
df['type'] = df['type'].fillna('Unknown')
# Fill using mode per type
df['type'] = df['type'].fillna('{Unassigned}')
df['comment'] = df['comment'].fillna('')
df['has_photo'] = df['photo'].notna().astype(int)
df['has_photo_after'] = df['photo_after'].notna().astype(int)
df['district'] = df['district'].str.title().fillna('Unknown')
df['subdistrict'] = df['subdistrict'].str.title().fillna('Unknown')
df['star'] = df['star'].fillna(-1).astype(int)
df['last_activity'] = df['last_activity'].fillna(df['timestamp'])
df['state'] = df['state'].fillna('unknown')
df['count_reopen'] = df['count_reopen'].fillna(0).astype(int)
```

## Feature Engineering
เพิ่มหลังหัวข้อมา

# fix or remove impossible broken rows

```python
print("Dataset shape before cleaning:", df.shape)
df = df.dropna(subset=['ticket_id', 'timestamp', 'lat', 'lon', 'organization'])
df = df[df['province'].str.lower() == 'กรุงเทพมหานคร'].copy()
df = df[df['count_reopen'].fillna(0) >= 0]
df = df.dropna(how='all')
df = df.reset_index(drop=True)
print("Dataset shape after cleaning:", df.shape)
print("Sample rows:")
df.head(40)
```

## Save Cleaned Data

```python
output_path = '../data_clean/bangkok_traffy_clean.csv'
import os
os.makedirs('../data_clean', exist_ok=True)
df.to_csv(output_path, index=False)
output_path
```
